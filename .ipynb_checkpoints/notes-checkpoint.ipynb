```python
import numpy as np
import matplotlib.pylab as plt
```

```python
size = 4
rank = 0
arr = np.zeros(5)

indices = np.zeros(size+1)

for i in range(size):
    indices[i] = 5 * rank / size
    print(str(indices[i]) + " ")
```

```python
from ipywidgets import interact,fixed
import ipywidgets as widgets

import math

def plotStartBlock(iteration, size):
    x = np.linspace(0, size, 1000)
    blockstart = ((size-x) // 2**(iteration-1)) * 2**(iteration-1)
    plt.plot(x, blockstart)


p = 16
for i in range(int(math.log2(p)), 0, -1):
    print(i)
    plotStartBlock(i, p)
    plt.show()
```
